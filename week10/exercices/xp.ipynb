 Exercise 1: Rewrite and Optimize a Vague Prompt

Scenario:
You’re an AI Prompt Engineer at a productivity startup. Your manager asks you to use ChatGPT to create a LinkedIn post promoting their new focus app, FlowNest. They send you this vague prompt:

    “Write something about productivity tips.”


Tasks

    Identify 3 critical issues with this prompt.
    (Hint: Think about brand voice, target users, and platform engagement.)

    Rewrite the prompt using all three techniques:
    Role prompting (e.g., “Act as a…”)
    Instruction-based prompting (e.g., “Give 3 actionable…”)

    Format/tone/length constraint (e.g., “Use bullet points” or “max 280 characters”)

    Add a final line that makes the prompt reusable (e.g., “Use this format whenever we promote productivity tools on social media”).


Critical Issue

Too vague / lacks specificity – The AI doesn’t know what type of productivity tips, the tone, or the focus app (FlowNest).

No target audience defined – LinkedIn posts need to engage professionals, but the prompt doesn’t clarify this.

No format or style guidance – Without direction, the output could be too long, too casual, or not scannable for social media.

Rewritten prompt:

“Act as a LinkedIn marketing expert for FlowNest, a focus and productivity app. 
Write a LinkedIn post that gives 3 actionable productivity tips for busy professionals. 
Use bullet points, maintain a professional yet engaging tone, and keep each tip under 280 characters. 
End with a call-to-action that subtly promotes FlowNest. 
Use this format whenever we promote productivity tools on social media.”

Exercise 2: Multi-Part Prompt for Quiz Generation

Scenario:
A 7th-grade science teacher uploads a short article on volcanic eruptions. You are tasked with writing a prompt for ChatGPT to create engaging educational content.


Tasks

    Write a complete prompt that asks the model to:
    Summarize the article in 2 simple bullet points
    Generate 3 age-appropriate multiple-choice questions, each with 1 correct and 2 distractors
    Include a friendly tone and simple vocabulary for 11–13-year-olds

    Output everything in a format suitable for Google Slides

    Compare your prompt to this vague one:

        “Make a quiz for kids about this article.”

Explain three key ways your version prevents irrelevant or low-quality output.

Summarize the article in 2 simple bullet points using easy-to-understand language.

Create 3 multiple-choice questions based on the article. Each question should:

Have 1 correct answer and 2 distractors.

Use age-appropriate vocabulary and a friendly tone.

Format everything for Google Slides so each slide can contain either a bullet-point summary or one question. Include a clear title for each slide.

Output example format:

Slide 1: Summary

Bullet point 1

Bullet point 2

Slide 2: Question 1
Q: [Insert question here]
A) [Option 1]
B) [Option 2]
C) [Option 3]
Correct answer: [Letter]

Slide 3: Question 2
…and so on for all 3 questions.

Make the tone friendly and fun, and the language simple enough for 11–13-year-olds.



“Make a quiz for kids about this article.”

Chatgpt : 7 quizs with 4 bullet points


1. Clear and structured instructions

The prompt specifies exactly what to produce: 2 bullet points for the summary, 3 multiple-choice questions, and a Google Slides-ready format.

By giving a specific output format (slide titles, question layout, correct answers), it reduces the chance of ChatGPT generating content that doesn’t fit the slides or is messy.

2. Audience and tone constraints

The prompt clearly states the target audience: 11–13-year-olds.

It asks for friendly language and simple vocabulary, which prevents content that is too technical, complicated, or inappropriate for middle school students.

3. Explicit content boundaries

Each multiple-choice question must include 1 correct answer and 2 distractors, which prevents ambiguous questions or too many options.

Asking for direct use in Google Slides focuses the output on concise, educational material, avoiding long essays, unrelated examples, or off-topic explanations.

General Prompt – E-commerce Customer Support Chatbot

Instructions for the Chatbot:

Greet the Customer:

"Hello [Customer Name]! Welcome to [Company Name]. How can I assist you today?"

Identify Customer Intent:
Determine if the inquiry is about:

Order Status → go to “Response: Order Status”

Returns or Exchanges → go to “Response: Returns/Exchanges”

Product Information → go to “Response: Product Info”

Payment or Billing Issues → go to “Response: Payment/Billing”

Other Questions → go to “Response: Escalation”

Responses by Scenario

1️⃣ Order Status

Ask for the order number:

"Could you please provide your order number?"

Provide tracking or estimated delivery:

"Thank you! Your order #[order number] is currently [status]. Estimated delivery is on [date]."

If there is a delay:

"It looks like there is a delay. You can track your package here: [link]."

2️⃣ Returns or Exchanges

Confirm product details:

"Which product would you like to return or exchange?"

Explain step-by-step process:

"Here’s how to proceed:

Fill out the return form here: [link].

Pack the product with the return label.

Drop it at a collection point or schedule a pickup."

Provide processing time:

"Once received, your refund or exchange will be processed within [X] business days."

3️⃣ Product Information

Provide accurate details:

"The [product name] is available in [colors, sizes]. The price is [price]."

Suggest alternatives if needed:

"If this model is out of stock, you can check these similar products: [link]."

4️⃣ Payment or Billing Issues

Clarify the problem:

"Could you specify if you’re experiencing a payment issue or a billing error?"

Respond accordingly:

"Your payment has been successfully processed. Here’s your receipt: [link]."

"An error was detected. I will submit a request for correction and keep you updated."

5️⃣ Other Questions / Escalation

If the inquiry doesn’t match a scenario:

"I will transfer your request to a human agent who can assist you promptly."

Provide details for the human agent:

"Customer Name: [Name], Order Number: [Order Number], Issue: [Description]."

6️⃣ Closure

Ask if they need further assistance:

"Is there anything else I can help you with today?"

End on a friendly note:

"Thank you for contacting [Company Name]! Have a great day."

✅ Key Features Matching the Structured Style:

Clear, labeled sections for each type of customer inquiry.

Step-by-step instructions and bullet points to ensure the chatbot follows a consistent workflow.

🌟 Exercise 6: Hallucination Spotting and Mitigation

Scenario:
You’re preparing a summary of a peer-reviewed article on climate change and marine biodiversity. ChatGPT inserts this claim:

    “Over 50% of marine species are projected to go extinct by 2050.”

This claim does not appear in the original article.


Tasks

1. Write a precise prompt instructing the model to avoid hallucinations and use only the provided text.

2. Rewrite the prompt using verification language, like:

    “Only include claims explicitly stated in the article.”
    “If uncertain, state that the information is not available.”

3. List 2 mitigation strategies you used to reduce hallucination risk.

4. Name 2 professional domains (e.g., legal, healthcare) where hallucinations are especially risky, and explain why.

1. Precise prompt to avoid hallucinations

“Summarize the following article on climate change and marine biodiversity. Only include information explicitly present in the text. Do not add claims, statistics, or conclusions that are not in the article.”

2. Prompt rewritten using verification language

“Summarize the provided article on climate change and marine biodiversity. Only include claims or data explicitly stated in the text. If a fact or statistic is not mentioned, indicate that the information is not available. Avoid inferring or adding any information not present in the article.”

3. Mitigation strategies used to reduce hallucination risk

Explicit source confinement: Instructing the model to rely solely on the provided text.

Verification fallback: Adding a rule to explicitly state “information not available” when the text doesn’t confirm a claim.

4. Professional domains with high hallucination risk

Healthcare / Medicine: Incorrect medical information can lead to misdiagnosis, improper treatment, or patient harm.

Legal / Compliance: Hallucinated legal claims, citations, or case references can lead to contractual errors, legal liabilities, or regulatory violations.
